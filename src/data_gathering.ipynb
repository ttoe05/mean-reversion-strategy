```python
import pandas as pd
import numpy as np
import yfinance as yf
```

```python
ticker = 'OSK'
df = yf.download(ticker, period="max", interval="1d", auto_adjust=False)
df.head()
```

```python
df.columns = ['Adj Close', 'Close', 'High', 'Low', 'Open', 'Volume']
df.head()
```

```python
# list of assets currently in portfolio
portfolio_assets = [
    'ACM',
    'OSK',
    'PATH',
    'RGLD',
    'AAAU',
    'CIBR',
    'IBIT',
    'ITA',
    'SGDM',
    'USO',
    'USXF',
    'XHE'
]

# iterate to get the data
for ticker in portfolio_assets:
    try:
        df_tmp = yf.download(ticker, period="max", interval="1d", auto_adjust=False)
        df_tmp.columns = ['Adj Close', 'Close', 'High', 'Low', 'Open', 'Volume']
        # save the data as a parquet file
        df_tmp.to_parquet(f'data/{ticker}.parquet')
    except Exception as e:
        print(f"Could not load ticker:\t{ticker}\n{e}")
```
